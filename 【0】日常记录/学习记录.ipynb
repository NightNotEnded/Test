## python学习记录


## 0.1  python的目录结构
### 0.1.1 环境变量

* 安装完python后，需要将在系统中添加环境变量
    * Python  主程序目录，默认是：`C:\Users\你的用户名\AppData\Local\Programs\Python\Python310\`
    * Python 3.10 脚本目录，默认是：`C:\Users\你的用户名\AppData\Local\Programs\Python\Python310\Scripts\`

* 配置环境：实际上就是系统的快捷程序清单
    * 例如 `python --version`，系统会在环境变量里面的目录寻找 `python.exe`，然后将 `--version`参数传递给它。完成前面内容
    * 给 `..\Scripts`也添加环境目录，因为这里会放置与python相关的命令行工具，例如：`pip.exe`,`jupyter.exe`等等（前提是通过pip下载的）


### 0.1.2 python包的结构
* `Scripts`：
    * 这里放置的都是python相关的命令行工具，例如：`pip.exe`,`jupyter.exe`等等（前提是通过pip下载的）
    * 当然，这个包还要添加环境变量


*  `bin`:
    * 在linux或macOS系统中会有，放置的是所有可执行命令，添加环境变量时候，添加它

* `lib`：
    * Python 标准库，文件和包都会放在这里
    * `lib\site-packages`
    * 如果是 `pip install xx`下载下来的包，会放在这里，如果它有命令行工具，就会放在 `Scripts`里面一个`.exe`文件
        *


### 0.1.3  pythonw
* pythonw.exe 是 "Python Windowed" 的缩写，它是一个不显示控制台窗口的 Python 解释器。
* 运行它后，不会显示控制台，适合运行那些GUI应用程序或后台脚本
